```python
import pandas as pd
import numpy as np

print("Đang đọc dữ liệu...")
df = pd.read_csv('nigerian_sales_data_5000_repeat_customers.csv')
df.columns = df.columns.str.lower().str.replace(' ', '_')
df['unit_price'] = df['unit_price'].replace('[\$,]', '', regex=True).astype(float)
df['total_sale_amount'] = df['total_sale_amount'].replace('[\$,]', '', regex=True).astype(float)
df['discount_applied'] = df['discount_applied'].fillna('0')

def calculate_actual_discount(row):
    discount_str = str(row['discount_applied'])
    if '%' in discount_str:
        percent_val = float(discount_str.replace('%', ''))
        return row['total_sale_amount'] * (percent_val / 100)
    elif '$' in discount_str:
        return float(discount_str.replace('$', ''))
    else:
        try:
            return float(discount_str)
        except:
            return 0.0


df['actual_discount_amount'] = df.apply(calculate_actual_discount, axis=1)
df.loc[df['return_status'] == 'Returned', 'total_sale_amount'] = 0.
df['date'] = pd.to_datetime(df['date'])

df = df.drop_duplicates()

output_file = 'cleaned_nigerian_sales_data.csv'
df.to_csv(output_file, index=False)

print(f"\nHoàn tất! Dữ liệu đã được làm sạch và lưu tại: {output_file}")
print("\nXem trước 5 dòng đầu tiên của dữ liệu đã làm sạch:")
print(df[['customer_id', 'total_sale_amount', 'actual_discount_amount', 'return_status']].head())
```

```python
import pandas as pd

df = pd.read_csv('cleaned_nigerian_sales_data.csv')
# Tách bảng Customers
df_customers = df[['customer_id', 'customer_name', 'gender', 'age']].drop_duplicates(subset=['customer_id'])
df_customers.to_csv('dim_customers.csv', index=False)

# Tách bảng Products
df_products = df[['product_id', 'product_name', 'category']].drop_duplicates(subset=['product_id'])
df_products.to_csv('dim_products.csv', index=False)

# Tách bảng Salespersons
df_salespersons = df[['salesperson_id', 'salesperson_name']].drop_duplicates(subset=['salesperson_id'])
df_salespersons.to_csv('dim_salespersons.csv', index=False)


df_transactions = df[['transaction_id', 'date', 'customer_id', 'product_id',
                      'salesperson_id', 'store_location', 'payment_method',
                      'quantity_sold', 'unit_price', 'discount_applied',
                      'actual_discount_amount', 'total_sale_amount', 'return_status']]
df_transactions.to_csv('fact_transactions.csv', index=False)

print("Đã tách thành công 4 file CSV!")
```
